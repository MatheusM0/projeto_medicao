```python
from ultralytics import YOLO

# 1. Carregar um modelo pré-treinado leve (YOLOv8 nano) para treino rápido
model = YOLO('yolov8n.pt')

# 2. Iniciar o treinamento
results = model.train(
    data='../data/raw/data.yaml',
    epochs=50,          # 50 épocas costumam ser suficientes para 2 classes simples
    imgsz=640,          # Resolução padrão
    batch=16,           # Tamanho do lote (reduza para 8 se faltar memória RAM/VRAM)
    name='medicao_madeira_v1'
)

print("Treinamento concluído! O modelo foi salvo na pasta 'runs/detect/medicao_madeira_v1/weights/best.pt'")

print("a")
```
